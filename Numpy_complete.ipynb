```python
# aggregation functions we can say summary 
# sum mean median max min var(variance) std(standard daviation) 
t = np.array([10,20,30,40,50,60,70,80,90])
print("sum : ",np.sum(t))
print("mean : ",np.mean(t))
print("median : ",np.median(t))
print("max : ",np.max(t))
print("min : ",np.min(t))
print("variance : ",np.var(t))
print("standard deviation : ",np.std(t))
```

```python
# broadcasting
prices = [100,200,300]
discount = 0.1  # 10% discount
final_discounted_prices = []

for price in prices:
    final_price = price - (price*discount)
    final_discounted_prices.append(final_price)
print(final_discounted_prices)
```

```python
# broadcasting- now what if we have million numbers of data then definitely using loop is a headache
prices = np.array([100,200,300])
discount = 0.1  # 10% discount
final_discounted_prices = prices - (prices*0.1)
print(final_discounted_prices)
```

```python
# broadcasting through single value , 2D array
prices = np.array([100,200,300])
result = prices * 2
print(result)
print("----"*10)

matrix = np.array([[1,2,3],[4,5,6]])
vector = np.array([10,20,30])
result = matrix + vector
print(result)
```

```python
# handling missing values 
# np.isnan --> detects missing value, detects if someething is dividing by 0 etc
# np.nan_to_num()
# np.isinf() --> to detect infinite value
arr = np.array([64,np.nan,70,np.nan,59,78])
print(arr)
print(np.isnan(arr))
```

```python
# replacing nan value using nan_to_num
clean_arr = np.nan_to_num(arr,nan = np.nanmean(arr))
print(clean_arr)
```

```python
# infinity np.isinf(arr)  finds is there any infinite value etc or not returns a boolean value 
arr1 = np.array([1,2,np.inf,3,-np.inf,10])
find_inf = np.isinf(arr1)
print(find_inf)
# replacing isinf value 
clean_arr = np.nan_to_num(arr1, posinf = 100, neginf= -100)
print(clean_arr)
```
